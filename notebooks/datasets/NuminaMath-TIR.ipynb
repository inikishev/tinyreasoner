```python
from tinyreasoner import chat_api

def load():
    local_path = hf_hub_download(
        repo_id="AI-MO/NuminaMath-TIR",
        filename="data/train-00000-of-00001.parquet",
        repo_type="dataset"
    )
    samples = []
    df = pl.read_parquet(local_path)

    for row in df.iter_rows(named=True):

        samples.append([
            chat_api.UserMessage(f'{row["problem"]}. Give a detailed step-by-step solution and the answer.'),
            chat_api.AssistantTurn(chat_api.Text(row["solution"]))
        ])


    return samples

load()[0]
```
