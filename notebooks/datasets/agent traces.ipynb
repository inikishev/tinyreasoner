```python
import polars as pl

import polars as pl
from huggingface_hub import hf_hub_download

def load():
    local_path = hf_hub_download(
        repo_id="trace-commons/agent-traces",
        filename="data/train-00000-of-00001.parquet",
        repo_type="dataset"
    )
    df = pl.concat([pl.read_parquet(local_path), pl.read_parquet(local_path)])
    
    examples = []
    for _, _, _, messages, tools, *_ in df.iter_rows():
        examples.append({"tools": tools, "messages": messages})
    return examples
        
examples = load()
print(len(examples))
examples[0]
```

```python
from tinyreasoner.tokenizer import TOKENIZER
success = 0
fail = 0
for example in examples:
    try:
        TOKENIZER.encode_chat(example)
        success += 1
    except Exception as e:
        print(e)
        fail += 1
        
success, fail
```
