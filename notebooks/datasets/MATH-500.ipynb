```python
def load():
    local_path = hf_hub_download(
        repo_id="HuggingFaceH4/MATH-500",
        filename="test.jsonl",
        repo_type="dataset"
    )


    df = pl.read_ndjson(local_path)

    chats = []
    for row in df.iter_rows(named=True):
        chats.append([
            chat_api.UserMessage(row['problem']),
            chat_api.AssistantMessage(row["answer"])
        ])

        chats.append([
            chat_api.UserMessage(f"{row['problem']}. Give the solution and the answer."),
            chat_api.AssistantMessage(f"### Solution\n{row['solution']}\n\n### Answer\n{row['answer']}")
        ])

    return chats

load()[0]
```
