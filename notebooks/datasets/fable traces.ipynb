```python
from tinyreasoner.tokenizer import TOKENIZER
from tinyreasoner import chat_api

def load():
    local_path = snapshot_download(
        repo_id="Glint-Research/Fable-5-traces",
        allow_patterns="pi-traces/*.jsonl",
        repo_type="dataset"
    )
    df = pl.read_ndjson(local_path)

    chats = []
    for item in df.iter_rows(named=True):
        if item["message"] is None: continue
        chats.append({"messages": item["message"]})

    samples = chat_api.convert_dataset(chats, True, True)
    return samples

samples = load()
print(len(samples))
samples[0]
```
